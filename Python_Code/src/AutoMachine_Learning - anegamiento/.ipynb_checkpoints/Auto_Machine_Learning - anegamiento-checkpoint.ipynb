# Definiendo el mejor modelo
Ahora que ya tengo las variables, procedo a probar y a definir cuál es el mejor modelo predicción usando metodología FIGS.
La metodología FIGS (Focused Identification of Germplasm Strategy) parte de un principio clave:

*Los ambientes donde fueron colectadas las accesiones explican, en parte, su adaptación a condiciones extremas (como sequía o exceso de agua).*

Para los modelos a usar, he definidio crear una variable objetivo llamada "resistente a sequía". Esta variable será de valor binario (1/0) y se genera a partir de los datos del ambiente histórico donde se colectó la accesión.
* **Resistente a sequía** = Accesiones colectadas en ambientes áridos o con baja precipitación y alta evapotranspiración.
* **No resistente** = Ambientes húmedos, con alta humedad o baja aridez.


# Esta parte por editar
En este caso las variables que usaré para crear la variable objetivo son
* 'bio12',  # Precipitación anual
* 'bio13',  # Precipitación del mes más húmedo
* 'bio14',  # Precipitación del mes más seco
* 'bio15',  # Estacionalidad de la precipitación
* 'bio18',  # Precipitación del trimestre más cálido
* 'bio19',  # Precipitación del trimestre más frío
* 'current_30arcsec_climaticMoistureIndex',  # Índice de humedad climática
* 'current_30arcsec_annualPET',              # Evapotranspiración potencial anual
* 'current_30arcsec_aridityIndexThornthwaite' # Índice de aridez de Thornthwait

## Paso 1 Carga del dataset ##

```python
# ==========================================
# CARGA Y REVISIÓN DEL DATASET
# ==========================================

# Cargar el archivo
df = pd.read_csv("./dataset_entrada/dataset_reducido_no_colineal.csv")

# Mostrar información general
print("Dimensiones:", df.shape)
print("\nPrimeras filas:")
display(df.head())

print("\nResumen de tipos de datos y nulos:")
display(df.info())
```

## Creando la variable a predecir ##
En este caso voy a predecir que accesiones que sean resistibles a anegamiento (Suelo saturado de agua).
Para este caso voy a realizar un agrupamiento con varias variables ambientales (K-Means) que me sirve para crear la variabe a predecir.

```python
# ============================================================
# Crear variable objetivo a partir de K-Means
# ============================================================

variables_humedad = [
    'bio12',  # Precipitación anual
    'bio13',  # Precipitación del mes más húmedo
    'bio14',  # Precipitación del mes más seco
    'bio15',  # Estacionalidad de la precipitación
    'bio18',  # Precipitación del trimestre más cálido
    'bio19',  # Precipitación del trimestre más frío
    'current_30arcsec_climaticMoistureIndex',  # Índice de humedad climática
    'current_30arcsec_annualPET',              # Evapotranspiración potencial anual
    'current_30arcsec_aridityIndexThornthwaite' # Índice de aridez de Thornthwaite
]


# Eliminar filas con valores faltantes en esas variables
df_cluster = df[variables_humedad].dropna()

#Escalamiento de las variables (muy importante para KMeans)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_cluster)

# Aplicar KMeans con 3 clusters
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
clusters = kmeans.fit_predict(X_scaled)

# Asignar el cluster al dataframe original (alineando los índices)
df.loc[df_cluster.index, 'cluster_anegamiento'] = clusters

# Calcular el promedio de humedad por cluster (para identificar el más húmedo)
cluster_means = df.groupby('cluster_anegamiento')[[
    'bio12', 'bio13', 'bio18', 'current_30arcsec_climaticMoistureIndex'
]].mean()

# Identificar el cluster más húmedo (mayor promedio de precipitación anual)
cluster_mas_humedo = cluster_means['bio12'].idxmax()

# Crear la variable binaria Resistencia_Anegamiento
df['Resistencia_Anegamiento'] = np.where(df['cluster_anegamiento'] == cluster_mas_humedo, 1, 0)

# Resumen de los resultados que he obtenido
print("Cluster más húmedo identificado:", cluster_mas_humedo)
print("\nPromedio por cluster (indicadores de humedad):")
print(cluster_means)
print("\nDistribución de Resistencia_Anegamiento:")
print(df['Resistencia_Anegamiento'].value_counts(normalize=True) * 100)

# (Opcional) Guardar el nuevo dataset
#df.to_csv("dataset_anegamiento.csv", index=False)
#print("\nArchivo guardado como 'dataset_anegamiento.csv'")

```

## Resultados de K-MEANS ##
Se forman tres grupos ambientales relacionados con la humedad y la precipitación, siendo el **grupo 1** el cluster menos húmedo y el **grupo 2** el cluster mas húmedo.

En la siguinete tabla se resume cómo se tomaron los valores de precipiotación para armar los clusters:

| cluster | bio12 | bio13 | bio18 | Índice humedad climática | Interpretación ecológica                                                            |
| ------- | ----- | ----- | ----- | ------------------------ | ----------------------------------------------------------------------------------- |
| 0       | −0.22 | +0.09 | +0.04 | −0.55                    | Ambientes intermedios (lluvia moderada, ligera aridez)                              |
| 1       | −0.49 | −0.73 | −0.68 | +0.23                    | Ambientes **más secos** (baja precipitación y humedad relativa)                     |
| 2       | +1.53 | +1.03 | +1.10 | +1.19                    | Ambientes **más húmedos** (mayor precipitación anual y persistencia de humedad) |

Antes de ejecutar el K-Means, se estandarizaron los valores de las variables a agrupar. Esto me permtite obtener valores entre -1 y 1. Los valores entre 0 y 1 representan condiciones mas humedas que el promedio. Los valores entre -1 y 0 representan las condicoones mas secas.

La distribución de resistencia a anegamiento quedó de la siguiente forma:

| Clase | Descripción                                        | Porcentaje  |
| ----- | -------------------------------------------------- | ----------- |
| 0     | Ambientes secos o intermedios → **no resistentes** | **82.48 %** |
| 1     | Ambientes húmedos → **resistentes**                | **17.52 %** |



## Eliminando Columnas ## 
Elimino las columnas que use para crear la variable objetivo. Esto lo hago porque al entrenar los modelos ya saben la respuesta antes de entrenar, porque la variable objetivo fue calculada directamente con esas columnas. Esto evita el data-lake
# ************************************************************************************************************************************************************

```python
#Variables a eliminar
variables = [
    'bio12',  # Precipitación anual
    'bio13',  # Precipitación del mes más húmedo
    'bio14',  # Precipitación del mes más seco
    'bio15',  # Estacionalidad de la precipitación
    'bio18',  # Precipitación del trimestre más cálido
    'bio19',  # Precipitación del trimestre más frío
    'current_30arcsec_climaticMoistureIndex',  # Índice de humedad climática
    'current_30arcsec_annualPET',              # Evapotranspiración potencial anual
    'current_30arcsec_aridityIndexThornthwaite' # Índice de aridez de Thornthwaite
]

# Convertir a H2OFrame
hf = h2o.H2OFrame(df)

y = "Resistencia_Anegamiento"

# Excluir también identificadores y el cluster
exclusiones = ['DOI', 'ACCENUMB', 'cluster_anegamiento'] + variables_humedad

X = [col for col in hf.columns if col not in exclusiones + [y]]

# Convertir la variable objetivo a categórica
hf[y] = hf[y].asfactor()

print(f"Número total de predictores: {len(X)}")
# Verificar dimensiones
hf[X].describe()
```

```python
# División de datos (70% entrenamiento / 30% prueba)
train, test = hf.split_frame(ratios=[0.7], seed=42)

# Separar predictores y variable objetivo (ajusta el nombre exacto)
y_col = "Resistencia_Anegamiento"
X_cols = [c for c in test.columns if c != y_col]
# Convertir a DataFrame de pandas
X_test = test[X_cols].as_data_frame()
y_test = test[y_col].as_data_frame()

# Crear carpeta de salida (si no existe)
import os
os.makedirs("./dataset_salida", exist_ok=True)

# Guardar los archivos CSV
X_test.to_csv("./dataset_salida/X_test_AML_anegamiento.csv", index=False)
y_test.to_csv("./dataset_salida/y_test_AML_anegamiento.csv", index=False)
```

```python
# Ejecución del AutoML
aml = H2OAutoML(
    max_models=20,              # número de modelos a entrenar
    seed=42,
    balance_classes=True,       # balancear clases (por la leve desproporción 82%-18%)
    sort_metric="AUC",          # métrica principal
    max_runtime_secs=600        # tiempo máximo (10 min aprox)
)

aml.train(x=X, y=y, training_frame=train)
```

```python
# Ranking de modelos
leaderboard = aml.leaderboard
leaderboard.head(rows=10)


```

```python
# Evaluación del modelo líder en conjunto de prueba
leader = aml.leader
perf = leader.model_performance(test_data=test)
perf
```

## Matriz de confusión del módelo ##



Este es el punto de probabilidad a partir del cual el modelo decide:

* Si la probabilidad ≥ 0.5631 entonces toma la clase 1 (Resistente a anegamiento)
* Si la probabilidad < 0.5631 entonces clase 0 (No resistente)

Lo anterior me permite deducir que el modelo necesita un poco más de 50% de confianza para clasificar como resistente.

| Métrica                   | Valor | Interpretación                                 |
| ------------------------- | ----- | ---------------------------------------------- |
| **Accuracy**              | 100%  | Predijo todas las muestras correctamente.      |
| **Precision**             | 100%  | No hubo falsos positivos.                      |
| **Recall (Sensibilidad)** | 100%  | No hubo falsos negativos.                      |
| **F1-Score**              | 1.0   | Máximo posible (precisión y recall perfectos). |


La matriz de confusión es perfecta lo que me lleva a dudar si pudo haber habido overfitting en el entrenamiento. Lo que se realiza entonces es una matriz de confusión para el test. Se obtiene lo siguiente:



De las métricas obtenidas se interpreta que:

| Métrica                                      | Valor  | Interpretación                                                                 |
